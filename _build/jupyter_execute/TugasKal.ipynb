# Gram-Schmidt Process: Finding q₂

## Given:
$$A = \begin{bmatrix} 7 & 12 & 3 & 9 \\ 15 & 2 & 18 & 6 \\ 4 & 11 & 8 & 14 \\ 10 & 5 & 16 & 1 \end{bmatrix}$$

## Step 1: q₁ (Completed) ✓
$$v_1 = a_1 = \begin{bmatrix} 7 \\ 15 \\ 4 \\ 10 \end{bmatrix}$$

$$\|v_1\| = \sqrt{7^2 + 15^2 + 4^2 + 10^2} = \sqrt{390}$$

$$q_1 = \frac{v_1}{\|v_1\|} = \frac{1}{\sqrt{390}}\begin{bmatrix} 7 \\ 15 \\ 4 \\ 10 \end{bmatrix}$$

---

## Step 2: Finding v₂

### Formula:
$$v_2 = a_2 - (a_2 \cdot q_1)q_1$$

### Compute projection coefficient:
$$a_2 \cdot q_1 = \frac{1}{\sqrt{390}}(12\cdot7 + 2\cdot15 + 11\cdot4 + 5\cdot10)$$
$$= \frac{1}{\sqrt{390}}(84 + 30 + 44 + 50) = \frac{208}{\sqrt{390}}$$

### Compute v₂:
$$v_2 = \begin{bmatrix} 12 \\ 2 \\ 11 \\ 5 \end{bmatrix} - \frac{208}{390}\begin{bmatrix} 7 \\ 15 \\ 4 \\ 10 \end{bmatrix}$$

$$v_2 = \begin{bmatrix} 12 \\ 2 \\ 11 \\ 5 \end{bmatrix} - \frac{104}{195}\begin{bmatrix} 7 \\ 15 \\ 4 \\ 10 \end{bmatrix}$$

$$v_2 \approx \begin{bmatrix} 8.267 \\ -6.000 \\ 8.867 \\ -0.333 \end{bmatrix}$$

---

## Step 3: Normalize to get q₂

### Compute ‖v₂‖:

$$\|v_2\| \approx \sqrt{68.34 + 36 + 78.62 + 0.11} \approx \sqrt{183.07} \approx 13.53$$
