# Tugas Matriks

## 1. Determinan Matriks A

Diketahui:

$$
A=
\begin{bmatrix}
2 & 1 & 3\\
0 & 4 & 5\\
1 & 2 & 1
\end{bmatrix}
$$

Metode Sarrus:

$$
\begin{bmatrix}
2 & 1 & 3 & 2 & 1\\
0 & 4 & 5 & 0 & 4\\
1 & 2 & 1 & 1 & 2
\end{bmatrix}
$$

### Diagonal Utama

$$
(2\times4\times1)+(1\times5\times1)+(3\times0\times2)
$$

$$
=8+5+0=13
$$

### Diagonal Sekunder

$$
(1\times4\times3)+(2\times5\times2)+(1\times0\times1)
$$

$$
=12+20+0=32
$$

### Determinan

$$
\det(A)=13-32=-19
$$

**Hasil:**

$$
\boxed{\det(A)=-19}
$$

## 3. Determinan Matriks C

Diketahui:

$$
C=
\begin{bmatrix}
1 & 2 & 3\\
2 & 4 & 6\\
1 & 1 & 1
\end{bmatrix}
$$

Metode Sarrus:

$$
\begin{bmatrix}
1 & 2 & 3 & 1 & 2\\
2 & 4 & 6 & 2 & 4\\
1 & 1 & 1 & 1 & 1
\end{bmatrix}
$$

### Diagonal Utama

$$
(1\times4\times1)+(2\times6\times1)+(3\times2\times1)
$$

$$
=4+12+6=22
$$

### Diagonal Sekunder

$$
(1\times4\times3)+(1\times6\times1)+(1\times2\times2)
$$

$$
=12+6+4=22
$$

### Determinan

$$
\det(C)=22-22=0
$$

**Hasil:**

$$
\boxed{\det(C)=0}
$$

Karena determinannya nol, maka matriks C merupakan matriks singular.

\end{bmatrix}

$$
(2 \times 2 \times 2)+(1 \times 1 \times 0)+(0 \times 1 \times 1)

$$
=0+2+2=4

Sehingga,

$$
\det(B)=8-4=4

# 9. Invers Matriks C

$$
C=
\begin{bmatrix}
3&0&2\\
2&0&-2\\
0&1&1
\end{bmatrix}
$$

Salin dua kolom pertama.

$$
\begin{bmatrix}

0&1&1&0&1
\end{bmatrix}
$$

## a. Diagonal Utama

$$
(3\times0\times1)+(0\times(-2)\times0)+(2\times2\times1)
$$

$$
=0+0+4=4
$$

## b. Diagonal Sekunder

$$
(0\times0\times2)+(1\times(-2)\times3)+(1\times2\times0)

$$
=0-6+0=-6
$$

Sehingga

$$
\det(C)=4-(-6)=10
